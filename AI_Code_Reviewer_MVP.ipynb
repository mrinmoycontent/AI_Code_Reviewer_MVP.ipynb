```python
# AI Code Reviewer - Temporary Reviewer

def review_code(code):
    return {
        "score": 85,
        "issues": [
            {
                "severity": "Low",
                "problem": "No input validation",
                "explanation": "The function assumes price and quantity are valid numeric values.",
                "suggested_fix": "Validate the inputs before performing the calculation."
            },
            {
                "severity": "Low",
                "problem": "Function could be simplified",
                "explanation": "The intermediate 'total' variable is not necessary.",
                "suggested_fix": "Return price * quantity directly."
            }
        ]
    }

review_result = review_code(code_to_review)

print("Code Quality Score:", review_result["score"])
print()

for issue in review_result["issues"]:
    print(f"[{issue['severity']}] {issue['problem']}")
    print("Explanation:", issue["explanation"])
    print("Suggested Fix:", issue["suggested_fix"])
    print("-" * 50)
```

```python
# ============================================================
# AI CODE REVIEWER MVP
# Complete Interactive Application
# ============================================================

import re
import html
import ipywidgets as widgets

from IPython.display import display, HTML, clear_output


# ============================================================
# 1. LANGUAGE SELECTOR
# ============================================================

language_dropdown = widgets.Dropdown(
    options=[
        "Python",
        "JavaScript",
        "Java",
        "C#",
        "C++",
        "SQL"
    ],
    value="Python",
    description="Language:",
    layout=widgets.Layout(width="350px")
)


# ============================================================
# 2. CODE INPUT
# ============================================================

code_box = widgets.Textarea(
    value="""def calculate_total(price, quantity):
    total = price * quantity
    return total

print(calculate_total(100, 2))""",

    placeholder="Paste your code here...",

    description="Code:",

    layout=widgets.Layout(
        width="100%",
        height="300px"
    )
)


# ============================================================
# 3. BUTTONS
# ============================================================

review_button = widgets.Button(
    description="🔍 Review Code",
    button_style="primary",
    layout=widgets.Layout(
        width="180px",
        height="45px"
    )
)

fix_button = widgets.Button(
    description="🛠️ Generate Fix",
    button_style="success",
    layout=widgets.Layout(
        width="180px",
        height="45px"
    )
)


# ============================================================
# 4. OUTPUT AREAS
# ============================================================

review_output = widgets.Output()

fix_output = widgets.Output()


# ============================================================
# 5. REVIEW DASHBOARD
# ============================================================

def show_review_dashboard(review_text, language):

    # --------------------------------------------------------
    # Find score
    # --------------------------------------------------------

    score_matches = re.findall(
        r'(?:Final Code Quality Score|Code Quality Score|Score)'
        r'.{0,40}?(\d{1,3})\s*(?:/100)?',
        review_text,
        re.IGNORECASE | re.DOTALL
    )

    if score_matches:
        score = int(score_matches[-1])
    else:
        score = 0

    score = max(0, min(score, 100))


    # --------------------------------------------------------
    # Rating
    # --------------------------------------------------------

    if score >= 80:
        rating = "Excellent"
    elif score >= 60:
        rating = "Good"
    elif score >= 40:
        rating = "Needs Improvement"
    else:
        rating = "Poor"


    # --------------------------------------------------------
    # Count severity
    # --------------------------------------------------------

    critical = len(
        re.findall(r'\bCritical\b', review_text, re.IGNORECASE)
    )

    high = len(
        re.findall(r'\bHigh\b', review_text, re.IGNORECASE)
    )

    medium = len(
        re.findall(r'\bMedium\b', review_text, re.IGNORECASE)
    )

    low = len(
        re.findall(r'\bLow\b', review_text, re.IGNORECASE)
    )


    # --------------------------------------------------------
    # Escape AI response
    # --------------------------------------------------------

    safe_review = html.escape(review_text)


    # --------------------------------------------------------
    # Dashboard
    # --------------------------------------------------------

    dashboard = f"""

    <div style="
        font-family: Arial, sans-serif;
        max-width: 950px;
        margin: 25px auto;
        padding: 28px;
        border: 1px solid #ddd;
        border-radius: 18px;
        background: white;
        box-shadow: 0 5px 20px rgba(0,0,0,0.08);
    ">

        <h1 style="
            margin: 0;
            font-size: 32px;
        ">
            🔍 AI Code Reviewer
        </h1>

        <p style="
            color: #666;
            margin-top: 8px;
        ">
            AI-powered code quality analysis
        </p>


        <div style="
            display: inline-block;
            padding: 7px 14px;
            background: #eef2ff;
            border-radius: 20px;
            margin: 10px 0 20px 0;
        ">
            💻 {language}
        </div>


        <!-- SCORE -->

        <div style="
            text-align: center;
            padding: 25px;
            margin: 10px 0 25px 0;
            border-radius: 15px;
            background: #f4f7ff;
        ">

            <div style="
                font-size: 55px;
                font-weight: bold;
            ">
                {score}/100
            </div>

            <div style="
                font-size: 21px;
                margin-top: 5px;
            ">
                {rating}
            </div>

        </div>


        <!-- SEVERITY CARDS -->

        <div style="
            display: flex;
            gap: 12px;
            flex-wrap: wrap;
            margin-bottom: 30px;
        ">


            <div style="
                flex: 1;
                min-width: 150px;
                padding: 16px;
                text-align: center;
                border: 1px solid #eee;
                border-radius: 12px;
            ">

                <div style="font-size: 22px;">🔴</div>

                <b>Critical</b>

                <div style="
                    font-size: 26px;
                    margin-top: 5px;
                ">
                    {critical}
                </div>

            </div>


            <div style="
                flex: 1;
                min-width: 150px;
                padding: 16px;
                text-align: center;
                border: 1px solid #eee;
                border-radius: 12px;
            ">

                <div style="font-size: 22px;">🟠</div>

                <b>High</b>

                <div style="
                    font-size: 26px;
                    margin-top: 5px;
                ">
                    {high}
                </div>

            </div>


            <div style="
                flex: 1;
                min-width: 150px;
                padding: 16px;
                text-align: center;
                border: 1px solid #eee;
                border-radius: 12px;
            ">

                <div style="font-size: 22px;">🟡</div>

                <b>Medium</b>

                <div style="
                    font-size: 26px;
                    margin-top: 5px;
                ">
                    {medium}
                </div>

            </div>


            <div style="
                flex: 1;
                min-width: 150px;
                padding: 16px;
                text-align: center;
                border: 1px solid #eee;
                border-radius: 12px;
            ">

                <div style="font-size: 22px;">🔵</div>

                <b>Low</b>

                <div style="
                    font-size: 26px;
                    margin-top: 5px;
                ">
                    {low}
                </div>

            </div>

        </div>


        <!-- REVIEW -->

        <h2>📋 Detailed AI Review</h2>

        <div style="
            white-space: pre-wrap;
            line-height: 1.65;
            padding: 22px;
            background: #fafafa;
            border: 1px solid #eee;
            border-radius: 12px;
            overflow-x: auto;
        ">{safe_review}</div>

    </div>
    """

    display(HTML(dashboard))


# ============================================================
# 6. REVIEW FUNCTION
# ============================================================

def review_code(button):

    with review_output:

        clear_output(wait=True)

        user_code = code_box.value.strip()
        language = language_dropdown.value


        if not user_code:

            print("⚠️ Please enter some code first.")
            return


        print("🔍 AI is reviewing your code...")
        print(f"Language: {language}")
        print()


        try:

            review_prompt = f"""

You are an expert software code reviewer.

Programming language:
{language}

Review the following {language} code:

--- CODE START ---

{user_code}

--- CODE END ---

Analyze the code for:

1. Bugs
2. Security issues
3. Code quality problems
4. Performance issues
5. Recommended improvements

For every genuine issue provide:

- Severity: Critical / High / Medium / Low
- Problem
- Explanation
- Suggested Fix

Do not invent problems.

If there are no issues in a category, clearly state that.

At the end provide:

Final Code Quality Score: X/100

Base the score on the actual quality of the submitted code.

"""


            response = client.models.generate_content(
                model="gemini-3.6-flash",
                contents=review_prompt
            )


            show_review_dashboard(
                response.text,
                language
            )


        except Exception as error:

            print("❌ Review failed.")
            print()
            print("Error:", error)


# ============================================================
# 7. GENERATE FIX FUNCTION
# ============================================================

def generate_fix(button):

    with fix_output:

        clear_output(wait=True)

        user_code = code_box.value.strip()
        language = language_dropdown.value


        if not user_code:

            print("⚠️ Please enter some code first.")
            return


        print("🛠️ AI is generating a fix...")
        print(f"Language: {language}")
        print()


        try:

            fix_prompt = f"""

You are an expert software developer.

Programming language:
{language}

Improve the following {language} code:

--- ORIGINAL CODE ---

{user_code}

--- END ORIGINAL CODE ---

Requirements:

1. Fix genuine bugs.
2. Improve security where necessary.
3. Improve code quality.
4. Add appropriate validation.
5. Add type annotations where useful.
6. Add documentation where useful.
7. Do not add unnecessary complexity.
8. Preserve the original purpose and behavior.
9. Do not invent problems.
10. Make sure the resulting code is syntactically valid {language}.

Return exactly these sections:

SUMMARY:

Explain what you changed.

FIXED CODE:

Provide the complete corrected code inside an appropriate code block.

CHANGES:

List the important changes and explain why they were made.

"""


            response = client.models.generate_content(
                model="gemini-3.6-flash",
                contents=fix_prompt
            )


            safe_fix = html.escape(response.text)


            display(HTML(f"""

            <div style="
                font-family: Arial, sans-serif;
                max-width: 950px;
                margin: 25px auto;
                padding: 28px;
                border: 1px solid #ddd;
                border-radius: 18px;
                background: white;
                box-shadow: 0 5px 20px rgba(0,0,0,0.08);
            ">

                <h2>
                    🛠️ AI Generated Fix
                </h2>

                <div style="
                    white-space: pre-wrap;
                    line-height: 1.65;
                    padding: 22px;
                    background: #fafafa;
                    border: 1px solid #eee;
                    border-radius: 12px;
                    overflow-x: auto;
                ">
{safe_fix}
                </div>

            </div>

            """))


        except Exception as error:

            print("❌ Fix generation failed.")
            print()
            print("Error:", error)


# ============================================================
# 8. CONNECT BUTTONS
# ============================================================

review_button.on_click(review_code)

fix_button.on_click(generate_fix)


# ============================================================
# 9. DISPLAY APPLICATION
# ============================================================

display(
    HTML("""
    <h1 style="
        font-family: Arial;
        margin-bottom: 5px;
    ">
        🤖 AI Code Reviewer MVP
    </h1>

    <p style="
        font-family: Arial;
        color: #666;
    ">
        Review, analyze and improve your code with AI.
    </p>
    """)
)

display(language_dropdown)

display(code_box)

display(
    widgets.HBox(
        [
            review_button,
            fix_button
        ],
        layout=widgets.Layout(
            margin="15px 0",
            gap="10px"
        )
    )
)

display(review_output)

display(fix_output)
```

```python
from google import genai
import getpass

api_key = getpass.getpass("Enter your Gemini API key: ")

client = genai.Client(api_key=api_key)

print("✅ Gemini client connected.")
```

```python
# ============================================================
# RESTORE AI CODE REVIEWER
# ============================================================

import re
import html
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

# ------------------------------------------------------------
# Language selector
# ------------------------------------------------------------

language_dropdown = widgets.Dropdown(
    options=[
        "Python",
        "JavaScript",
        "Java",
        "C#",
        "C++",
        "SQL"
    ],
    value="Python",
    description="Language:",
    layout=widgets.Layout(width="350px")
)

# ------------------------------------------------------------
# Code input
# ------------------------------------------------------------

code_box = widgets.Textarea(
    value="""def calculate_total(price, quantity):
    total = price * quantity
    return total

print(calculate_total(100, 2))""",
    placeholder="Paste your code here...",
    description="Code:",
    layout=widgets.Layout(
        width="100%",
        height="300px"
    )
)

# ------------------------------------------------------------
# Buttons
# ------------------------------------------------------------

review_button = widgets.Button(
    description="🔍 Review Code",
    button_style="primary",
    layout=widgets.Layout(
        width="180px",
        height="45px"
    )
)

generate_fix_button = widgets.Button(
    description="🛠️ Generate Fix",
    button_style="success",
    layout=widgets.Layout(
        width="200px",
        height="45px"
    )
)

# ------------------------------------------------------------
# Outputs
# ------------------------------------------------------------

review_output = widgets.Output()
fix_output = widgets.Output()

# ------------------------------------------------------------
# Review function
# ------------------------------------------------------------

def review_code(button):

    with review_output:

        clear_output(wait=True)

        user_code = code_box.value.strip()
        language = language_dropdown.value

        if not user_code:
            print("⚠️ Please enter some code first.")
            return

        print("🔍 AI is reviewing your code...")
        print(f"Language: {language}")
        print()

        prompt = f"""
You are an expert software code reviewer.

Programming language: {language}

Review this {language} code:

--- CODE START ---
{user_code}
--- CODE END ---

Identify:

1. Bugs
2. Security issues
3. Code quality problems
4. Performance issues
5. Recommended improvements

For every genuine issue provide:

Severity: Critical / High / Medium / Low
Problem:
Explanation:
Suggested Fix:

Do not invent problems.

Finally provide:

Final Code Quality Score: X/100
"""

        try:

            response = client.models.generate_content(
                model="gemini-3.6-flash",
                contents=prompt
            )

            text = response.text

            scores = re.findall(
                r'(?:Final Code Quality Score|Code Quality Score)'
                r'.{0,30}?(\d{1,3})',
                text,
                re.IGNORECASE | re.DOTALL
            )

            score = int(scores[-1]) if scores else 0
            score = max(0, min(score, 100))

            if score >= 80:
                rating = "Excellent"
            elif score >= 60:
                rating = "Good"
            elif score >= 40:
                rating = "Needs Improvement"
            else:
                rating = "Poor"

            critical = len(re.findall(r'\bCritical\b', text, re.I))
            high = len(re.findall(r'\bHigh\b', text, re.I))
            medium = len(re.findall(r'\bMedium\b', text, re.I))
            low = len(re.findall(r'\bLow\b', text, re.I))

            display(HTML(f"""
            <div style="
                font-family:Arial;
                max-width:950px;
                margin:20px auto;
                padding:25px;
                border:1px solid #ddd;
                border-radius:16px;
                background:white;
            ">

            <h1>🔍 AI Code Reviewer</h1>

            <p>Language: <b>{html.escape(language)}</b></p>

            <div style="
                text-align:center;
                padding:25px;
                background:#f4f7ff;
                border-radius:14px;
            ">
                <div style="font-size:52px;font-weight:bold;">
                    {score}/100
                </div>
                <div style="font-size:20px;">
                    {rating}
                </div>
            </div>

            <br>

            <div style="display:flex;gap:10px;flex-wrap:wrap;">

                <div style="flex:1;padding:15px;text-align:center;border:1px solid #eee;border-radius:10px;">
                    🔴<br><b>Critical</b><br>
                    <span style="font-size:24px;">{critical}</span>
                </div>

                <div style="flex:1;padding:15px;text-align:center;border:1px solid #eee;border-radius:10px;">
                    🟠<br><b>High</b><br>
                    <span style="font-size:24px;">{high}</span>
                </div>

                <div style="flex:1;padding:15px;text-align:center;border:1px solid #eee;border-radius:10px;">
                    🟡<br><b>Medium</b><br>
                    <span style="font-size:24px;">{medium}</span>
                </div>

                <div style="flex:1;padding:15px;text-align:center;border:1px solid #eee;border-radius:10px;">
                    🔵<br><b>Low</b><br>
                    <span style="font-size:24px;">{low}</span>
                </div>

            </div>

            <h2>📋 Detailed AI Review</h2>

            <div style="
                white-space:pre-wrap;
                line-height:1.6;
                padding:20px;
                background:#fafafa;
                border-radius:10px;
            ">
{html.escape(text)}
            </div>

            </div>
            """))

        except Exception as e:

            print("❌ Review failed.")
            print(type(e).__name__)
            print(e)


# ------------------------------------------------------------
# Generate Fix function
# ------------------------------------------------------------

def generate_fix(button):

    with fix_output:

        clear_output(wait=True)

        user_code = code_box.value.strip()
        language = language_dropdown.value

        if not user_code:
            print("⚠️ Please enter some code first.")
            return

        print("🛠️ AI is generating a fix...")
        print(f"Language: {language}")
        print()

        prompt = f"""
You are an expert {language} software developer.

Improve this {language} code:

--- CODE START ---
{user_code}
--- CODE END ---

Fix genuine bugs and improve code quality.

Do not change the purpose of the program.
Do not add unnecessary complexity.

Return:

SUMMARY:
Explain the improvements.

FIXED CODE:
Provide the complete corrected code.

CHANGES:
List the important changes.
"""

        try:

            response = client.models.generate_content(
                model="gemini-3.6-flash",
                contents=prompt
            )

            print("✅ Fix generated successfully!")
            print()

            display(HTML(f"""
            <div style="
                font-family:Arial;
                max-width:950px;
                margin:20px auto;
                padding:25px;
                border:1px solid #ddd;
                border-radius:16px;
                background:white;
            ">

            <h2>🛠️ AI Generated Fix</h2>

            <div style="
                white-space:pre-wrap;
                line-height:1.6;
                padding:20px;
                background:#fafafa;
                border-radius:10px;
            ">
{html.escape(response.text)}
            </div>

            </div>
            """))

        except Exception as e:

            print("❌ Fix generation failed.")
            print(type(e).__name__)
            print(e)


# ------------------------------------------------------------
# Connect buttons
# ------------------------------------------------------------

review_button.on_click(review_code)
generate_fix_button.on_click(generate_fix)


# ------------------------------------------------------------
# Display application
# ------------------------------------------------------------

display(HTML("""
<h1 style="font-family:Arial;">
🤖 AI Code Reviewer MVP
</h1>

<p style="font-family:Arial;color:#666;">
Review, analyze and improve your code with AI.
</p>
"""))

display(language_dropdown)
display(code_box)

display(
    widgets.HBox([
        review_button,
        generate_fix_button
    ])
)

display(review_output)
display(fix_output)

print("✅ AI Code Reviewer restored successfully.")
```
